```python
import seaborn as sns
import matplotlib.pyplot as plt

df_titanic = sns.load_dataset('titanic')


```

```python
sns.histplot(
  x='sex', hue='survived', stat='count',
  multiple='stack', shrink=.6,
  data=df_titanic)
```

```python
order = df_titanic.groupby("class", observed=True)["survived"].mean().sort_values(ascending=False).index
sns.barplot(data=df_titanic, x="survived", y="class", order=order, errorbar=None)
```
